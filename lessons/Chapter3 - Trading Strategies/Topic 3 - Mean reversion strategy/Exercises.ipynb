# Exercises — Mean reversion strategy (RSI)

[DataCamp exercise](https://campus.datacamp.com/courses/financial-trading-in-python/trading-strategies?ex=8) · see `Notes.md` in this folder for the summary.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Locate the project's data folder regardless of where this notebook runs from
DATA = next(p / "course materials" / "data"
            for p in [Path.cwd(), *Path.cwd().parents]
            if (p / "course materials" / "data").is_dir())

def load(name):
    """Load an OHLCV CSV with a parsed DatetimeIndex."""
    return pd.read_csv(DATA / name, index_col="Date", parse_dates=True)

def price(name, col, year=None):
    """Single-asset price DataFrame (column = `col`) for use with bt."""
    df = load(name)
    if year:
        df = df[df.index.year == year]
    return df["Close"].rename(col).to_frame()
```

```python
import bt
import talib
import numpy as np

data = price("AMZN-stock-data.csv", "AMZN", year=2020)
rsi = talib.RSI(data["AMZN"], timeperiod=14)

# Buy the dip (RSI<30 -> +1), sell the peak (RSI>70 -> -1), else flat
signal = pd.DataFrame(0.0, index=data.index, columns=["AMZN"])
signal["AMZN"] = np.where(rsi < 30, 1.0, np.where(rsi > 70, -1.0, 0.0))

bt_strategy = bt.Strategy("RSI_MeanReversion", [
    bt.algos.WeighTarget(signal),
    bt.algos.Rebalance(),
])
bt_result = bt.run(bt.Backtest(bt_strategy, data))
bt_result.plot(title="RSI mean reversion")
plt.show()
```
