```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Locate the project's data folder regardless of where this notebook runs from
DATA = next(p / "course materials" / "data"
            for p in [Path.cwd(), *Path.cwd().parents]
            if (p / "course materials" / "data").is_dir())

def load(name):
    """Load an OHLCV CSV with a parsed DatetimeIndex."""
    return pd.read_csv(DATA / name, index_col="Date", parse_dates=True)

def price(name, col, year=None):
    """Single-asset price DataFrame (column = `col`) for use with bt."""
    df = load(name)
    if year:
        df = df[df.index.year == year]
    return df["Close"].rename(col).to_frame()
```

```python
import talib

amzn = load("AMZN-stock-data.csv")
amzn = amzn[amzn.index.year == 2020]
amzn["RSI"] = talib.RSI(amzn["Close"], timeperiod=14)
print(amzn[["Close", "RSI"]].tail())
```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(9, 6), sharex=True)
ax1.plot(amzn.index, amzn["Close"]); ax1.set_title("AMZN price")
ax2.plot(amzn.index, amzn["RSI"], color="orange")
ax2.axhline(70, color="red", ls="--")           # overbought
ax2.axhline(30, color="green", ls="--")          # oversold
ax2.set_title("RSI (14)"); ax2.set_ylim(0, 100)
plt.show()
```
